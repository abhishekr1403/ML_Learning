```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv('canada_per_capita_income.csv', names=['year','income'], header=0 )
df.shape

```

```python
%matplotlib inline
plt.scatter(df.year, df.income, color='red', marker='*')
```

```python
model = linear_model.LinearRegression()
model.fit(df[["year"]], df.income)
```

```python
model.predict([[2020]])
```
