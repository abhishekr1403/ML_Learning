```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model

```

```python
df = pd.read_csv('homeprices.csv')
df
```

```python
%matplotlib inline
plt.xlabel('Area')
plt.ylabel('Price')
plt.scatter(df.area, df.price, color='red', marker='+')
plt.show()
```

```python
feature = df.drop('price', axis='columns')
feature
```

```python
model = linear_model.LinearRegression()
model.fit(feature, df['price'])
```

```python
model.predict([[3300]])
```

```python
model.coef_
```

```python
model.intercept_
```

```python
df_new = pd.read_csv('areas.csv')
df_new
```

```python
p = model.predict(df_new)
p
```

```python
df_new['price'] = p
df_new
```

```python
df_new.to_csv('prediction.csv', index=False)
```
